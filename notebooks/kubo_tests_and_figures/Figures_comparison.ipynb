```julia
using LinearAlgebra
function sqrt_psd(A::AbstractMatrix{<:Complex})
    H = A#(A + A')/2              # enforce Hermitian numerically
    #H = A
    F = eigen(H)                           # Hermitian eigendecomposition
    λ = F.values#@. max(F.values, 0.0)              # clamp tiny negatives to 0
    U = F.vectors
    return U * Diagonal(sqrt.(λ)) * U'     # √A
end
function fidelity(ρ::AbstractMatrix{<:Complex}, σ::AbstractMatrix{<:Complex})
    # ρs = sqrt_psd(ρ)
    # σs = sqrt_psd(σ)
    # s = svdvals(ρs * σs)
    # return real((sum(s))^2)
    ρs = sqrt_psd(ρ)                  # √ρ
    X  = ρs * σ * ρs                           # √ρ σ √ρ
    Y  = sqrt_psd(X)                   # √(√ρ σ √ρ)
    return real(tr(Y))^2
end
```
